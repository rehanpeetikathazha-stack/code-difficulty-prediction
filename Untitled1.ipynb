```python
import pandas as pd

FEATURES = ["LOC", "nesting_depth", "loops", "conditions",
            "functions", "variables", "cyclomatic_complexity"]
EXPERIENCE_ORDER = ["Less than 6 months", "6 to 12 months", "1 to 2 years", "More than 2 years"]   # NEW

# ---------- 1. Load ----------
snip = pd.read_csv("snippets.csv")
part = pd.read_csv("participants.csv", dtype={"participant_code": str})
rate = pd.read_csv("ratings.csv", dtype={"participant_code": str})
print("Loaded:", len(snip), "snippets |", len(part), "students |", len(rate), "ratings")

# ---------- 2. Look at the basics ----------
print(rate.dtypes)
print(rate.isna().sum())

# ---------- 3. Cleaning checks (count first, then remove) ----------
log = {}

bad_missing = rate["difficulty_1to10"].isna()
log["missing difficulty"] = bad_missing.sum()

bad_range = ~rate["difficulty_1to10"].between(1, 10) & ~bad_missing
log["difficulty outside 1-10"] = bad_range.sum()

bad_conf = rate["confidence_1to5"].notna() & ~rate["confidence_1to5"].between(1, 5)
log["confidence outside 1-5"] = bad_conf.sum()

bad_code = ~rate["participant_code"].str.fullmatch(r"\d{4}", na=False)
log["participant code not 4 digits"] = bad_code.sum()

keys = ["participant_code", "snippet_id", "is_repeat"]
bad_dup = rate.duplicated(subset=keys, keep="first")      # keeps the first submission
log["duplicate submissions"] = bad_dup.sum()

# NEW: ratings from students who are missing in the participants table
bad_student = ~rate["participant_code"].isin(part["participant_code"])
log["student not in participants table"] = bad_student.sum()

print("\nRows that would be removed:")
for k, v in log.items():
    print(f"  {k}: {v}")

n_before = len(rate)
rate = rate[~(bad_missing | bad_range | bad_conf | bad_code | bad_dup | bad_student)].copy()
print("Ratings before:", n_before, "| after:", len(rate), "| removed:", n_before - len(rate))

# Unknown snippet IDs
unknown = ~rate["snippet_id"].isin(snip["snippet_id"])
print("Ratings with unknown snippet ID:", unknown.sum())
rate = rate[~unknown]

# NEW: check the participant table itself
print("\nParticipants with an unexpected experience value:",
      (~part["programming_experience"].isin(EXPERIENCE_ORDER)).sum())
print("Duplicate participant codes:", part["participant_code"].duplicated().sum())

# ---------- 4. Repeat-snippet reliability ----------
first = rate[rate["is_repeat"] == "N"]
again = rate[rate["is_repeat"] == "Y"]
merged = first.merge(again, on=["participant_code", "snippet_id"], suffixes=("_first", "_repeat"))
merged["gap"] = (merged["difficulty_1to10_repeat"] - merged["difficulty_1to10_first"]).abs()
print("\nStudents with a repeat pair:", len(merged))
print("Mean gap:", round(merged["gap"].mean(), 2), "| max gap:", merged["gap"].max(),
      "| students with gap > 3:", (merged["gap"] > 3).sum())
print(merged["gap"].value_counts().sort_index())

# ---------- 5. Build the modelling table (code features only: UNCHANGED) ----------
target = (first.groupby("snippet_id")["difficulty_1to10"]
               .mean().rename("mean_rating").reset_index())
model_table = snip[["snippet_id"] + FEATURES].merge(target, on="snippet_id", how="left")

print("\nModelling table shape:", model_table.shape)
print("Missing values:", model_table.isna().sum().sum())
assert len(model_table) == len(snip), "Row count changed"
assert model_table.isna().sum().sum() == 0, "Missing values remain"

model_table.to_csv("model_table.csv", index=False)
print(model_table.head())

# ---------- 6. NEW: rating-level table (one row per student x snippet) ----------
rating_level = (first
    .merge(part[["participant_code", "programming_experience", "python_comfort_1to5"]],
           on="participant_code", how="left", validate="m:1")
    .merge(snip[["snippet_id"] + FEATURES], on="snippet_id", how="left", validate="m:1"))

# experience as an ordered number: 0 = less than 6 months ... 3 = more than 2 years
rating_level["experience_level"] = pd.Categorical(
    rating_level["programming_experience"], categories=EXPERIENCE_ORDER, ordered=True).codes
assert (rating_level["experience_level"] >= 0).all(), "Unknown experience level found"
assert rating_level[["difficulty_1to10", "experience_level"] + FEATURES].isna().sum().sum() == 0

print("\nRating-level table shape:", rating_level.shape,
      "| students:", rating_level["participant_code"].nunique(),
      "| snippets:", rating_level["snippet_id"].nunique())
print("Students per experience level:")
print(rating_level.drop_duplicates("participant_code")["programming_experience"]
                  .value_counts().reindex(EXPERIENCE_ORDER))

rating_level.to_csv("rating_level_table.csv", index=False)

# ---------- 7. NEW: confidence summary per snippet (a separate result, NOT a predictor) ----------
snippet_conf = (first.groupby("snippet_id")
                     .agg(mean_difficulty=("difficulty_1to10", "mean"),
                          mean_confidence=("confidence_1to5", "mean"))
                     .reset_index())
snippet_conf.to_csv("snippet_confidence.csv", index=False)
print("\nSaved: model_table.csv, rating_level_table.csv, snippet_confidence.csv")
```
