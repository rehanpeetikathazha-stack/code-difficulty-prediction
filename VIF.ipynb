```python
import pandas as pd, numpy as np
import statsmodels.api as sm
from statsmodels.stats.outliers_influence import variance_inflation_factor

df = pd.read_csv("real_cleaned_dataset.csv")
CODE = ["LOC", "nesting_depth", "loops", "conditions", "functions", "variables", "cyclomatic_complexity"]

def vif_table(data, cols):
    X = sm.add_constant(data[cols].astype(float))     # the constant column is required
    return pd.Series({c: variance_inflation_factor(X.values, i)
                      for i, c in enumerate(X.columns) if c != "const"}).round(2)

snip = df.drop_duplicates("snippet_id")               # one row per snippet for the CODE features
print(vif_table(snip, CODE))

# rank check: include the constant column, or the exact dependence is missed
Xc = sm.add_constant(snip[CODE].astype(float))
print("rank:", np.linalg.matrix_rank(Xc.values), "of", Xc.shape[1])
```

```python
import pandas as pd
import numpy as np

df = pd.read_csv("real_cleaned_dataset.csv")

features = [
    "LOC",
    "nesting_depth",
    "loops",
    "conditions",
    "functions",
    "variables",
    "cyclomatic_complexity"
]

# X should contain only predictor columns
X = df[features].copy()

print("Unique values in each feature:")
print(X.nunique())

print("\nCorrelation matrix:")
print(X.corr().round(3))

print("\nFeatures with zero variance:")
print(X.columns[X.nunique() <= 1].tolist())

print("\nExact duplicate columns:")
print([
    (a, b)
    for i, a in enumerate(X.columns)
    for b in X.columns[i + 1:]
    if X[a].equals(X[b])
])
```
